Credits and Keywords

```python
import ast

# Convert JSON-like strings to Python objects
credits['cast'] = credits['cast'].apply(ast.literal_eval)
credits['crew'] = credits['crew'].apply(ast.literal_eval)
keywords['keywords'] = keywords['keywords'].apply(ast.literal_eval)

# Get only the top 3 cast members
credits['cast'] = credits['cast'].apply(lambda x: [i['name'] for i in x][:3])

# Get director from crew
def get_director(crew):
    for person in crew:
        if person['job'] == 'Director':
            return person['name']
    return np.nan

credits['director'] = credits['crew'].apply(get_director)
credits = credits.drop('crew', axis=1)

# Get keyword names
keywords['keywords'] = keywords['keywords'].apply(lambda x: [i['name'] for i in x])
```

## Merge Everything till now

```python
#converting id to int to match other tables
metadata['id']=pd.to_numeric(metadata['id'],errors='coerce')
metadata=metadata.dropna(subset=['id'])
metadata['id']=metadata['id'].astype(int)

#merge
movies=metadata.merge(credits,on='id').merge(keywords,on='id')

#preview
movies[['title', 'overview','cast','director','keywords']].head()
```

## Build a Content-Based Recommendation Engine

Creating a "Soup" of Metadata for Each Movie


```python
# Fill NaNs with empty strings
movies['overview'] = movies['overview'].fillna('')
movies['keywords'] = movies['keywords'].apply(lambda x: ' '.join(x))
movies['cast'] = movies['cast'].apply(lambda x: ' '.join(x))
movies['director'] = movies['director'].fillna('')

# Combine into a single soup
def create_soup(x):
    return x['overview'] + ' ' + x['keywords'] + ' ' + x['cast'] + ' ' + x['director']

movies['soup'] = movies.apply(create_soup, axis=1)
```

```python
indices = pd.Series(movies.index, index=movies['title'].str.lower())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the vectorizer
tfidf = TfidfVectorizer(stop_words='english')

# Fill any NaN values in the 'soup' column
movies['soup'] = movies['soup'].fillna('')

# Create the TF-IDF matrix
tfidf_matrix = tfidf.fit_transform(movies['soup'])
```

```python
from sklearn.neighbors import NearestNeighbors

# Create the KNN model
model = NearestNeighbors(metric='cosine', algorithm='brute')
model.fit(tfidf_matrix)
```

Vectorizing the Text Using CountVector


```python
def get_recommendations(title, n=10):
    # Make title lowercase for consistent matching
    title = title.lower()

    # Check if title is in our index mapping
    if title not in indices:
        return f"❌ Movie '{title}' not found in the database. Please check the spelling."

    # Get the index of the movie
    idx = indices[title]

    # Compute the nearest neighbors
    distances, indices_found = model.kneighbors(tfidf_matrix[idx], n_neighbors=n + 1)

    # Fetch recommended movies (skip the first one, which is the movie itself)
    recommendations = movies.iloc[indices_found[0][1:]][['title']]

    # Reset index and make it start from 1
    recommendations = recommendations.reset_index(drop=True)
    recommendations.index += 1

    return recommendations
```
